```python
import pandas as pd
import numpy as np
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt

REPO = "../.." 
FIG_DIR = "../figures"
import os
os.makedirs(FIG_DIR, exist_ok=True)
```

```python
PHASES = [
    ("Phase 1", "Gewichts- vs.\nAktivierungs-\nDekomposition", "#4C72B0"),
    ("Phase 2", "Random-Init-\nKontrolle", "#DD8452"),
    ("Phase 3", "Trace-Drift &\nRekonziliation", "#55A868"),
    ("Phase 3.5", "Fusion vs.\nQuantisierung", "#C44E52"),
    ("Phase 4", "Prädiktor-Test\n(Kernanalyse)", "#8172B2"),
    ("Phase 5", "Spike-\nSelection", "#937860"),
    ("Phase 6", "Architektonische\nUrsachensuche", "#DA8BC3"),
]
```

```python
fig, ax = plt.subplots(figsize=(7.0, 2.6), layout="constrained")
n = len(PHASES)
box_w, box_h, gap = 1.0, 0.75, 0.25
x_positions = [i * (box_w + gap) for i in range(n)]

for i, (label, desc, color) in enumerate(PHASES):
    x = x_positions[i]
    is_core = label == "Phase 4"
    lw = 2.2 if is_core else 1.0
    ax.add_patch(plt.Rectangle((x, 0), box_w, box_h, facecolor=color, alpha=0.25,
                                edgecolor=color, linewidth=lw))
    ax.text(x + box_w / 2, box_h * 0.72, label, ha="center", va="center",
            fontsize=8, fontweight="bold" if is_core else "normal")
    ax.text(x + box_w / 2, box_h * 0.32, desc, ha="center", va="center", fontsize=6.2)
    if i < n - 1:
        ax.annotate("", xy=(x_positions[i + 1], box_h / 2), xytext=(x + box_w, box_h / 2),
                    arrowprops=dict(arrowstyle="->", color="gray", lw=1.0))

ax.set_xlim(-0.2, x_positions[-1] + box_w + 0.2)
ax.set_ylim(-0.15, box_h + 0.15)
ax.axis("off")
ax.set_title("Sechs-Phasen-Analyse-Pipeline (Phase 4 = Kernanalyse, AXIOM-Paper)", fontsize=8.5)

fig.savefig(f"{FIG_DIR}/fig_02_pipeline_overview.pdf")
fig.savefig(f"{FIG_DIR}/fig_02_pipeline_overview.png", dpi=200)
plt.close(fig)
```
