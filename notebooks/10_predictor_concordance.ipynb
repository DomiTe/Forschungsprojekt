# 10 -- Predictor Concordance (Fig. 1 Appendix)

Pairwise Spearman correlation between the three sensitivity-score *values* themselves
(independent of the loss-damage target), for all 6 model x dataset combinations, PTQ.
Reproduces Table 9 of the report (Sec. 5.6).

Source CSV: `results/review_response/csv/predictor_concordance.csv`


```python
# Requirements: pandas==3.0.5, numpy==2.5.1, matplotlib==3.11.1, seaborn==0.13.2, scipy==1.18.0
# All notebooks in this report use the same environment; paths below are relative to
# report/notebooks/, so the notebook must be run with its own directory as the working
# directory (the default for `jupyter nbconvert --execute` and for Jupyter's own kernel).
import pandas as pd
import numpy as np
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt

REPO = "../.."  # report/notebooks -> report -> repo root
FIG_DIR = "../figures"
import os
os.makedirs(FIG_DIR, exist_ok=True)
```

```python
conc = pd.read_csv(f"{REPO}/results/review_response/csv/predictor_concordance.csv")
conc
```

Pivot from long (one row per pair) to wide (one row per model x dataset, one column per pair) to match the report table layout.


```python
MODEL_LABEL = {"cnn": "CNN", "resnet18_no_weights": "ResNet-18", "resnet50_no_weights": "ResNet-50"}
DATASET_LABEL = {"IMAGENET100": "ImageNet100", "CIFAR10": "CIFAR10"}
PAIR_LABEL = {
    "raw_trh_vs_dwsq": "rho(Sraw,Spert)",
    "raw_trh_vs_trh_dwsq": "rho(Sraw,Shawq)",
    "dwsq_vs_trh_dwsq": "rho(Spert,Shawq)",
}

wide = conc.pivot(index=["model", "dataset"], columns="pair", values="rho").reset_index()
wide = wide.rename(columns=PAIR_LABEL)
wide["Modell"] = wide["model"].map(MODEL_LABEL)
wide["Datensatz"] = wide["dataset"].map(DATASET_LABEL)

table9 = wide[["Modell", "Datensatz", "rho(Sraw,Spert)", "rho(Sraw,Shawq)", "rho(Spert,Shawq)"]]
table9 = table9.sort_values(["Modell", "Datensatz"]).reset_index(drop=True)
for c in table9.columns[2:]:
    table9[c] = table9[c].round(2)
table9.to_csv(f"{FIG_DIR}/tab_09_predictor_concordance.csv", index=False)
table9
```
