```python
import pandas as pd
import numpy as np
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt

REPO = "../.." 
FIG_DIR = "../figures"
import os
os.makedirs(FIG_DIR, exist_ok=True)
```

```python
kfac = pd.read_csv(f"{REPO}/results/20260816_230437_38678/csv/spike_layer_kfac.csv")
kfac.head()
```

```python
TARGETS = [
    ("resnet50_no_weights", "conv1"),
    ("resnet50_no_weights", "fc"),
    ("resnet18_no_weights", "layer1.1.conv1"),
    ("cnn", "conv1"),
    ("cnn", "conv2"),
]
MODEL_LABEL = {"cnn": "CNN", "resnet18_no_weights": "ResNet-18", "resnet50_no_weights": "ResNet-50"}

rows = []
for model, layer in TARGETS:
    sub = kfac[(kfac["model"] == model) & (kfac["dataset"] == "CIFAR10")
               & (kfac["variant"] == "ptq") & (kfac["layer"] == layer)
               & (kfac["seed"] == "aggregate")]
    r = sub.iloc[0]
    ratio = r["measured_per_param"] / r["predicted_per_param"]
    rows.append({
        "Modell": MODEL_LABEL[model], "Schicht": layer,
        "Vorhergesagt": r["predicted_per_param"], "Gemessen": r["measured_per_param"],
        "Verhältnis": ratio, "log10(Verhältnis)": np.log10(ratio),
    })

table8 = pd.DataFrame(rows)
table8.to_csv(f"{FIG_DIR}/tab_08_kfac_decomposition.csv", index=False)
table8
```
